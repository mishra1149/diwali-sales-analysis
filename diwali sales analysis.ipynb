# EXPLORATORY DATA ANALYSIS

## GENDER

```python
df.columns
```

```python
ax=sns.countplot(x='Gender' ,data=df)
for bars in ax.containers:
    ax.bar_label(bars)
```

```python
sales_gen=df.groupby(['Gender'],as_index=False)['Amount'].sum().sort_values(by='Amount',ascending=False)
 # df.groupby('Group_Column')['Target_Column'].aggregation_function
```

```python
sns.barplot(x='Gender',y='Amount',data=sales_gen)
```

### ANALYSIS

#### from above graph we can see that most of the buyers are female and even the purchasing power of female  are greater than man

# AGE

```python
sns.countplot(x='Age Group',hue='Gender',data=df)
```

```python
ax=sns.countplot(x='Age Group',hue='Gender',data=df)
for bars in ax.containers:
    ax.bar_label(bars)
```

```python
#total amount vs age group
sales_age=df.groupby(['Age Group','Gender'],as_index=False)['Amount'].sum().sort_values(by='Amount',ascending=False)
sales_age
```

```python
ax=sns.barplot(x='Age Group',y='Amount',hue='Gender',data=sales_age)
```

# STATE

```python
# total mo of order from top 10 rajya
df.columns
```

```python
top_10=df.groupby(['rajya'],as_index=False)['Orders'].sum().sort_values(by='Orders',ascending=False).head(10)
top_10
```

```python
plt.figure(figsize=(12, 6))
ax = sns.barplot(y='rajya', x='Orders', data=top_10, palette='viridis')

plt.title('Total Number of Orders from Top 10 Rajya', fontsize=15)
plt.xlabel('Rajya (States)', fontsize=12)
plt.ylabel('Total Orders', fontsize=12)
```

#### df.groupby( 'किस आधार पर ग्रुप करना है' )[ 'जिसका टोटल/एवरेज निकालना है' ].sum()


```python
# total amount/sales from top 10 rajya
sales_state=df.groupby(['rajya'], as_index=False)['Amount'].sum().sort_values(by="Amount",ascending=False).head(10)
sales_state
```

```python
sns.set(rc={'figure.figsize':(15,5)})
sns.barplot(data=sales_state,x='rajya',y='Amount')
```

## ANALYSIS

### FROM ABOVE GRAPH WE CAN SEE THAT UNEXPETEDLY MOST OF THE ORDER ARE FROM UTTARPRADESH MAHARASHTRA KARNATAKA RESP.

# OCCUPATION

```python
sns.set(rc={'figure.figsize':(20,5)})
ax=sns.countplot(data=df,x='Occupation')
for bars in ax.containers:
    ax.bar_label(bars)
```

```python
sales_data=df.groupby(['Occupation'],as_index=False)['Amount'].sum().sort_values(by='Amount',ascending=False)
sns.set(rc={'figure.figsize':(20,5)})
sns.barplot(data=sales_data,x='Occupation',y='Amount')
```

### from above graph we see that most of the buyers are from IT,healthcare and aviation

# PRODUCT CATEGORY

```python
sns.set(rc={'figure.figsize':(25,5)})
ax=sns.countplot(data=df,x='Product_Category')
for bars in ax.containers:
    ax.bar_label(bars)
```

```python
sales_category=df.groupby(['Product_Category'],as_index=False)['Amount'].sum().sort_values(by='Amount',ascending=False)
sns.set(rc={'figure.figsize':(30,5)})
sns.barplot(data=sales_category,x='Product_Category',y='Amount')
```

### from above graph  we can see that most of the sold products are from food clothing and electronic 

```python
sales_state1=df.groupby(['Product_ID'],as_index=False)['Orders'].sum().sort_values(by='Orders',ascending=False).head(10)
sns.set(rc={'figure.figsize':(20,5)})
sns.barplot(data=sales_state1,x='Product_ID',y='Orders')
```
